### datatype conversion

```python
df['Order.Date']=pd.to_datetime(df['Order.Date'])
df['Ship.Date']=pd.to_datetime(df['Ship.Date'])
```

```python
df['Order.ID'].duplicated().sum()
```

### Order_Year, Order_DayName, Order_Month, Order_Quarter Extracted by Order.Date

```python
df['Order_year']=df['Order.Date'].dt.year
df['Order_year']=df['Order.Date'].dt.day_name()
df['Order_year']=df['Order.Date'].dt.month
df['Order_year']=df['Order.Date'].dt.quarter
```

## Rename Columns

### calculate shipping_days column

```python
df['Shipping_Days']=df['Ship.Date']-df['Order.Date']
```

### order_value column calculate karo using 
#### Sales < 100 → Low Sales 100–500 → Medium Sales > 500 → High

```python
def sales_category(Sales):
    if Sales<100:
        return 'Low'
    elif Sales<=500:
        return 'Medium'
    else:
        return 'High'

df['Order_Value']=df['Sales'].apply(sales_category)
```

### profit_status column calculate karo using 
#### Profit > 0 → Profit, Profit = 0 → Break-even, Profit < 0 → Loss

```python
def profit_status(Profit):
    if Profit>0:
        return 'Profit'
    elif Profit<0:
        return 'Loss'
    else:
        return 'Break-even'

df['Profit_Status']=df['Profit'].apply(profit_status)
```

### discout_category column calculate karo using
#### 0 → No Discount, 0–20% → Low Discount,20–50% → Medium Discount, 50% → High Discount

```python
def discount_category(discount):
    if discount<=20:
        return 'Low Discount'
    elif discount<=50:
        return 'Medium Discount'
    elif discount>50:
        return 'High Discount'
    else:
        return 'No Discount'

df['Discount_Category']=df['Discount'].apply(discount_category)
```

### calculate profit_margin column using (Profit / Sales) × 100

```python
df["Profit_Margin"] = np.where(
    df["Sales"] != 0,
    (df["Profit"] / df["Sales"]) * 100,
    0
)
```

### save cleaned_superstore.csv file

```python
df.to_csv('cleaned_superstore.csv', index=False)
print('File saved successfully...')
```
